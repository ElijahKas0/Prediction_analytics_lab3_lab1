## Введение

В данной лабораторной работе реализуется двухэтапная система анализа метеорологических временных рядов для шести городов России: Москва, Санкт-Петербург, Сочи, Геленджик, Благовещенск и Находка. Используются архивные метеоданные за период 2019–2025 гг., сохранённые в формате `.parquet`. Каждый город представляет определённую климатическую зону, что позволяет исследовать задачу как классификации климатических режимов, так и прогнозирования погодных параметров в различных климатических условиях.

Работа состоит из двух основных этапов.

**Этап 1 — классификация временных рядов.**  
По окну метеорологических наблюдений фиксированной длины определяется, к какому городу и типу климата относится временной ряд. Для решения задачи используются методы анализа временных рядов и специализированные модели классификации временных последовательностей.

**Этап 2 — прогнозирование температуры.**  
Для каждого города строится отдельная модель прогнозирования температуры воздуха на 30 дней вперёд. Далее классификация и прогнозирование объединяются в единый пайплайн: система сначала определяет климатический тип входного временного ряда, а затем автоматически выбирает соответствующую модель прогноза.

В рамках работы выполняются:

- разведочный анализ метеоданных;
- исследование сезонности, трендов и выбросов;
- анализ стационарности временных рядов;
- построение признаков для задач классификации и прогнозирования;
- обучение и сравнение моделей классификации;
- обучение и сравнение моделей прогнозирования;
- оценка качества моделей с использованием статистических метрик и тестов;
- интеграция всех компонентов в единый программный пайплайн.

Для предотвращения утечки данных используется временное разделение выборки:

| Период | Назначение |
|--------|------------|
| 2019–2023 гг. | обучение (train) |
| 2024 г. | валидация (validation) |
| 2025 г. | тестирование (test) |

Основной рабочей частотой анализа выбрана дневная частота наблюдений. Исходные данные загружаются из годовых `.parquet`-файлов, после чего при необходимости выполняется ресемплинг и унификация временной сетки.

```python
DATA_DIR = Path("data")

CITIES: List[str] = [
    "Москва",
    "Санкт-Петербург",
    "Сочи",
    "Геленджик",
    "Благовещенск",
    "Находка",
]

CLIMATE_GROUPS: Dict[str, str] = {
    "Москва": "умеренно-континентальный",
    "Санкт-Петербург": "умеренно-континентальный",
    "Сочи": "субтропический",
    "Геленджик": "субтропический",
    "Благовещенск": "муссонно-континентальный",
    "Находка": "муссонно-континентальный",
}

YEARS: List[int] = list(range(2019, 2026))

RAW_FEATURES: List[str] = [
    "temperature_2m",
    "relative_humidity_2m",
    "precipitation",
    "rain",
    "snowfall",
    "weathercode",
    "wind_speed_10m",
    "surface_pressure",
]

TRAIN_YEARS = [2019, 2020, 2021, 2022, 2023]
VAL_YEARS = [2024]
TEST_YEARS = [2025]

WINDOW_DAYS = 30
FORECAST_HORIZON = 30

print("CITIES:", CITIES)
print("CLIMATES:", sorted(set(CLIMATE_GROUPS.values())))
print(f"Window for classification: {WINDOW_DAYS} days")
print(f"Forecast horizon: {FORECAST_HORIZON} days")
```

```python
# Загрузка parquet-файлов

def load_city_hourly(city: str,
                     data_dir: Path = DATA_DIR,
                     years: List[int] = YEARS) -> pd.DataFrame:

    city_parts = []

    for year in years:

        filename = f"{city}_{year}-01-01_{year}-12-31.parquet"
        filepath = data_dir / filename

        if not filepath.exists():
            print(f"[warn] Файл не найден: {filepath}")
            continue

        df = pd.read_parquet(filepath)

        # Нормализация времени

        if "time" in df.columns:
            df["time"] = pd.to_datetime(df["time"])
            df = df.set_index("time")

        elif "date" in df.columns:
            df["date"] = pd.to_datetime(df["date"])
            df = df.set_index("date")

        else:
            df.index = pd.to_datetime(df.index)

        df = df.sort_index()

        # оставляем только ожидаемые признаки
        cols = [c for c in RAW_FEATURES if c in df.columns]
        df = df[cols]

        city_parts.append(df)

    if len(city_parts) == 0:
        raise FileNotFoundError(f"Не найдено parquet-файлов для города: {city}")

    # Склейка всех лет

    full = pd.concat(city_parts).sort_index()

    # удаляем дубли времени
    full = full[~full.index.duplicated(keep="first")]

    # Проверка частоты

    inferred_freq = pd.infer_freq(full.index[:100])

    # Если данные уже дневные → оставляем как есть
    # Если часовые → приводим к регулярной сетке
    if inferred_freq in ["H", "h"]:

        full = full.asfreq("h")

    elif inferred_freq in ["D", None]:

        # Создаём полную дневную сетку
        idx = pd.date_range(
            start=full.index.min(),
            end=full.index.max(),
            freq="D"
        )

        full = full.reindex(idx)

    full.index.name = "time"

    return full
```

```python
def load_all_hourly(cities: List[str] = CITIES) -> Dict[str, pd.DataFrame]:

    data = {}

    for city in cities:

        print(f"Загрузка: {city}")

        city_df = load_city_hourly(city)

        data[city] = city_df

        print(
            f"  shape={city_df.shape}, "
            f"period={city_df.index.min().date()} → {city_df.index.max().date()}"
        )

    return data


hourly: Dict[str, pd.DataFrame] = load_all_hourly()
```

```python
summary = pd.DataFrame({

    "rows": {
        city: len(df)
        for city, df in hourly.items()
    },

    "start": {
        city: df.index.min()
        for city, df in hourly.items()
    },

    "end": {
        city: df.index.max()
        for city, df in hourly.items()
    },

    "missing_temperature": {
        city: int(df["temperature_2m"].isna().sum())
        for city, df in hourly.items()
    },

    "missing_%": {
        city: round(df.isna().mean().mean() * 100, 2)
        for city, df in hourly.items()
    },

})

summary
```

```python
def resample_to_daily(df: pd.DataFrame) -> pd.DataFrame:

    freq = pd.infer_freq(df.index[:50])
    
    if freq in ["D", None]:

        daily = df.copy()

        if "temperature_2m" in daily.columns:

            daily["temperature_min"] = daily["temperature_2m"]
            daily["temperature_max"] = daily["temperature_2m"]
            daily["temperature_range"] = 0.0

        daily.index.name = "date"

        return daily
    
    # Агрегация часовых данных
    
    agg_mean = [
        "temperature_2m",
        "relative_humidity_2m",
        "surface_pressure",
        "wind_speed_10m",
    ]

    agg_sum = [
        "precipitation",
        "rain",
        "snowfall",
    ]

    daily_mean = df[agg_mean].resample("D").mean()

    daily_sum = df[agg_sum].resample("D").sum(min_count=1)

    def _mode(x: pd.Series):

        x = x.dropna()

        if len(x) == 0:
            return np.nan

        return float(x.mode().iloc[0])

    daily_code = (
        df["weathercode"]
        .resample("D")
        .apply(_mode)
        .rename("weathercode")
    )

    daily_tmin = (
        df["temperature_2m"]
        .resample("D")
        .min()
        .rename("temperature_min")
    )

    daily_tmax = (
        df["temperature_2m"]
        .resample("D")
        .max()
        .rename("temperature_max")
    )

    daily_trange = (
        daily_tmax - daily_tmin
    ).rename("temperature_range")

    daily = pd.concat([
        daily_mean,
        daily_sum,
        daily_code,
        daily_tmin,
        daily_tmax,
        daily_trange,
    ], axis=1)

    daily.index.name = "date"

    return daily
```

```python
# Формирование ежедневных-данных

daily: Dict[str, pd.DataFrame] = {

    city: resample_to_daily(df)

    for city, df in hourly.items()

}
```

```python
# Проверка итогового набора

preview = pd.DataFrame({

    "days": {
        city: len(df)
        for city, df in daily.items()
    },

    "nan_%": {
        city: round(df.isna().mean().mean() * 100, 3)
        for city, df in daily.items()
    },

    "mean_temperature": {
        city: round(df["temperature_2m"].mean(), 2)
        for city, df in daily.items()
    },

    "total_precipitation": {
        city: round(df["precipitation"].sum(), 1)
        for city, df in daily.items()
    },

})

preview
```

### §2.1.5. Анализ выбросов

Экстремальные значения в метеоданных могут соответствовать:

- аномальной жаре
- циклонам
- ураганам
- сильным осадкам

##### Выявление экстремальных температур

```python
outlier_summary = []

for city, df in daily_clean.items():

    temp = df["temperature_2m"]

    z = np.abs(stats.zscore(temp))

    n_outliers = int((z > 3).sum())

    outlier_summary.append({
        "city": city,
        "outliers_temperature": n_outliers
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary)
```

##### Визуализация экстремумов

```python
city = "Москва"

df = daily_clean[city]

z = np.abs(stats.zscore(df["temperature_2m"]))

outliers = df[z > 3]

plt.figure(figsize=(15, 4))

plt.plot(df.index,
         df["temperature_2m"],
         lw=0.8,
         label="Температура")

plt.scatter(outliers.index,
            outliers["temperature_2m"],
            color="red",
            s=20,
            label="Выбросы")

plt.title(f"Температурные аномалии: {city}")
plt.legend()
plt.show()
```

### §2.1.6. Анализ стационарности

Для проверки стационарности используется тест Дики-Фуллера

##### Функция ADF-теста

```python
def adf_test(series, title=""):

    series = series.dropna()

    result = adfuller(series)

    print(f"ADF test: {title}")
    print(f"ADF statistic : {result[0]:.4f}")
    print(f"p-value       : {result[1]:.6f}")

    if result[1] < 0.05:
        print("Ряд стационарен")
    else:
        print("Ряд нестационарен")

    print("-" * 50)
```

##### Проверка температуры

```python
for city in daily_clean.keys():

    adf_test(
        daily_clean[city]["temperature_2m"],
        city
    )
```

##### Дифференцирование рядов

Для устранения тренда и сезонности используем дифференцирование.

```python
diff_data = {}

for city, df in daily_clean.items():

    tmp = df.copy()

    tmp["temperature_diff"] = (
        tmp["temperature_2m"]
        .diff()
    )

    diff_data[city] = tmp
```

```python
for city in diff_data.keys():

    adf_test(
        diff_data[city]["temperature_diff"].dropna(),
        f"{city} (diff)"
    )
```

##### Сравнение исходного и дифференцированного ряда

- сезон

##### Создание временных признаков

```python
def add_time_features(df: pd.DataFrame) -> pd.DataFrame:

    out = df.copy()

    out["year"] = out.index.year
    out["month"] = out.index.month
    out["day"] = out.index.day

    out["dayofweek"] = out.index.dayofweek

    # Сезон
    season_map = {
        12: "winter",
        1: "winter",
        2: "winter",

        3: "spring",
        4: "spring",
        5: "spring",

        6: "summer",
        7: "summer",
        8: "summer",

        9: "autumn",
        10: "autumn",
        11: "autumn",
    }

    out["season"] = out["month"].map(season_map)

    return out
```

```python
feature_data = {}

for city, df in daily_clean.items():

    feature_data[city] = add_time_features(df)

feature_data["Москва"].head()
```

### §2.2.2. Циклические признаки

Месяц и день недели являются циклическими величинами:

- декабрь близок к январю
- воскресенье близко к понедельнику

Поэтому используются sin/cos-преобразования

##### Циклическое кодирование

```python
def add_cyclic_features(df: pd.DataFrame) -> pd.DataFrame:

    out = df.copy()

    # Месяц
    out["month_sin"] = np.sin(2 * np.pi * out["month"] / 12)
    out["month_cos"] = np.cos(2 * np.pi * out["month"] / 12)

    # День недели
    out["dow_sin"] = np.sin(2 * np.pi * out["dayofweek"] / 7)
    out["dow_cos"] = np.cos(2 * np.pi * out["dayofweek"] / 7)

    return out
```

```python
for city in feature_data:

    feature_data[city] = add_cyclic_features(feature_data[city])

feature_data["Сочи"].head()
```

### §2.2.3. Лаговые признаки

Метеорологические процессы обладают выраженной временной инерцией

Для температуры и давления создаются лаги:

- 1 день
- 3 дня
- 7 дней
- 14 дней
- 30 дней

```python
LAGS = [1, 3, 7, 14, 30]

LAG_FEATURES = [
    "temperature_2m",
    "surface_pressure",
    "relative_humidity_2m",
]

def add_lag_features(df: pd.DataFrame) -> pd.DataFrame:

    out = df.copy()

    for col in LAG_FEATURES:

        for lag in LAGS:

            out[f"{col}_lag_{lag}"] = out[col].shift(lag)

    return out
```

```python
for city in feature_data:

    feature_data[city] = add_lag_features(feature_data[city])

feature_data["Москва"].head()
```

### §2.2.4. Скользящие статистики

Скользящие окна позволяют учитывать:

- краткосрочные погодные процессы
- недельные колебания
- устойчивые синоптические режимы

Используются окна:

- 3 дня
- 7 дней
- 14 дней
- 30 дней

```python
ROLL_WINDOWS = [3, 7, 14, 30]

ROLL_FEATURES = [
    "temperature_2m",
    "surface_pressure",
    "precipitation",
]

def add_rolling_features(df: pd.DataFrame) -> pd.DataFrame:

    out = df.copy()

    for col in ROLL_FEATURES:

        for w in ROLL_WINDOWS:

            out[f"{col}_roll_mean_{w}"] = (
                out[col]
                .rolling(w)
                .mean()
            )

            out[f"{col}_roll_std_{w}"] = (
                out[col]
                .rolling(w)
                .std()
            )

    return out
```

```python
for city in feature_data:

    feature_data[city] = add_rolling_features(feature_data[city])

feature_data["Москва"].head()
```

### §2.2.5. Признаки динамики

Дополнительно вычисляются:

- скорость изменения температуры
- ускорение изменения температуры
- темп изменения давления

##### Производные временных рядов

```python
def add_dynamic_features(df: pd.DataFrame) -> pd.DataFrame:

    out = df.copy()

    # Скорость изменения температуры
    out["temp_velocity"] = (
        out["temperature_2m"]
        .diff()
    )

    # Ускорение температуры
    out["temp_acceleration"] = (
        out["temp_velocity"]
        .diff()
    )

    # Изменение давления
    out["pressure_change"] = (
        out["surface_pressure"]
        .diff()
    )

    return out
```

```python
for city in feature_data:

    feature_data[city] = add_dynamic_features(feature_data[city])

feature_data["Санкт-Петербург"].head()
```

### §2.2.6. Климатическая норма

Климатическая норма — среднее значение температуры для данного дня года по всем годам наблюдений

```python
climate_norms = {}

for city, df in feature_data.items():

    norm = (
        df.groupby(df.index.dayofyear)["temperature_2m"]
        .mean()
    )

    climate_norms[city] = norm
```

```python
for city in feature_data:

    df = feature_data[city].copy()

    df["dayofyear"] = df.index.dayofyear

    df["climate_norm_temp"] = (
        df["dayofyear"]
        .map(climate_norms[city])
    )

    # Температурная аномалия
    df["temp_anomaly"] = (
        df["temperature_2m"] -
        df["climate_norm_temp"]
    )

    feature_data[city] = df
```

### §2.2.7. Агрегированные признаки для классификации

Для задачи классификации используются окна фиксированной длины.

Из каждого окна извлекаются:

- средняя температура
- амплитуда температуры
- количество дождливых дней
- средняя влажность
- максимальный ветер
- стандартное отклонение температуры

```python
WINDOW_DAYS = 30

def build_window_features(window: pd.DataFrame) -> dict:

    feats = {}

    temp = window["temperature_2m"]

    feats["temp_mean"] = temp.mean()
    feats["temp_std"] = temp.std()

    feats["temp_min"] = temp.min()
    feats["temp_max"] = temp.max()

    feats["temp_amplitude"] = (
        temp.max() - temp.min()
    )

    feats["precip_days"] = (
        window["precipitation"] > 0
    ).sum()

    feats["precip_mean"] = (
        window["precipitation"]
        .mean()
    )

    feats["humidity_mean"] = (
        window["relative_humidity_2m"]
        .mean()
    )

    feats["wind_max"] = (
        window["wind_speed_10m"]
        .max()
    )

    feats["pressure_std"] = (
        window["surface_pressure"]
        .std()
    )

    return feats
```

```python
classification_rows = []

for city, df in feature_data.items():

    for i in range(WINDOW_DAYS, len(df)):

        window = df.iloc[i - WINDOW_DAYS:i]

        feats = build_window_features(window)

        feats["city"] = city
        feats["date"] = df.index[i]

        classification_rows.append(feats)

classification_df = pd.DataFrame(classification_rows)

print(classification_df.shape)
classification_df.head()
```

## §2.3. Построение моделей классификации временных рядов

В данной секции реализуется первый этап двухэтапной системы — классификация климатического типа по временному ряду метеопараметров за фиксированное окно наблюдений

### Теоретическое обоснование выбора моделей

Задача классификации метеорологических временных рядов обладает рядом особенностей:

- высокая сезонность
- сильная временная корреляция наблюдений
- различия между климатами проявляются не только в абсолютных значениях температуры, но и в динамике
- наличие шумов, выбросов и экстремальных погодных явлений
- схожесть некоторых климатов (например, Москва ↔ Санкт-Петербург)

Поэтому обычные табличные методы без временной структуры могут терять важную информацию о форме временного ряда.

### Какие модели плохо подходят для задачи

#### 1. Обычная линейная регрессия

Не подходит, поскольку:

- не учитывает временную структуру
- не моделирует нелинейные зависимости
- чувствительна к мультиколлинеарности признаков

#### 2. K-Means / методы кластеризации

Не подходят как основная модель, поскольку:

- задача является supervised learning
- кластеры не обязаны совпадать с климатическими группами
- временная динамика теряется

#### 3. Классический RandomForest на «сырых» временных рядах

Без специального преобразования временного ряда модель:

- плохо воспринимает порядок наблюдений
- не извлекает форму временной динамики
- требует ручного feature engineering

#### 4. Полносвязные нейросети (MLP)

Для небольшой выборки метеоданных:

- склонны к переобучению
- требуют большого объёма данных
- хуже специализированных TSC-подходов

### Выбранные модели классификации

Для исследования используются две модели:

| Модель | Назначение |
|--------|------------|
| LogisticRegression | табличный baseline |
| MiniROCKET + LogisticRegression | специализированная TSC-модель |

### Почему выбран MiniROCKET

MiniROCKET относится к современным методам Time Series Classification (TSC):

- работает непосредственно с временными рядами
- использует набор случайных свёрточных ядер
- извлекает форму и динамику сигнала
- очень быстро обучается
- показывает state-of-the-art результаты на задачах TSC

Для метеоданных это особенно важно, поскольку климат определяется:

- сезонной структурой
- амплитудами
- динамикой температуры
- поведением влажности и давления

```python
# Для классификации используются скользящие окна длиной

WINDOW_DAYS = 30

# Каждое окно содержит:

# температуру
# влажность;
# давление
# скорость ветра
# осадки

# Целевая переменная — город
```

```python
TS_FEATURES = [
    "temperature_2m",
    "relative_humidity_2m",
    "surface_pressure",
    "wind_speed_10m",
    "precipitation",
]

WINDOW_DAYS = 30

X_windows = []
y_windows = []
window_dates = []

for city in CITIES:

    df_city = daily_clean[city].copy()

    for i in range(WINDOW_DAYS, len(df_city)):

        window = df_city.iloc[i - WINDOW_DAYS:i]

        X_windows.append(window[TS_FEATURES].values)

        y_windows.append(city)

        window_dates.append(df_city.index[i])

X_windows = np.array(X_windows, dtype=np.float32)
y_windows = np.array(y_windows)

print("Форма массива окон:", X_windows.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_encoded = le.fit_transform(y_windows)

print("Классы:")
print(le.classes_)
```

##### Разделение train / validation / test

Используется временное разделение:

| Период | Назначение |
|--------|------------|
| 2019–2023 | train |
| 2024 | validation |
| 2025 | test |

```python
window_dates = pd.to_datetime(window_dates)

train_mask = window_dates.year <= 2023
val_mask = window_dates.year == 2024
test_mask = window_dates.year == 2025

X_train = X_windows[train_mask]
X_val = X_windows[val_mask]
X_test = X_windows[test_mask]

y_train = y_encoded[train_mask]
y_val = y_encoded[val_mask]
y_test = y_encoded[test_mask]

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

##### Baseline-модель: Logistic Regression

Для baseline-подхода преобразуем временные окна в агрегированные статистики

```python
def aggregate_window_features(window):

    feats = []

    for j in range(window.shape[1]):

        col = window[:, j]

        feats.extend([
            np.mean(col),
            np.std(col),
            np.min(col),
            np.max(col),
            np.median(col),
        ])

    return feats


X_train_tab = np.array([aggregate_window_features(w) for w in X_train])
X_val_tab = np.array([aggregate_window_features(w) for w in X_val])
X_test_tab = np.array([aggregate_window_features(w) for w in X_test])

print(X_train_tab.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logreg = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(
        max_iter=5000,
        multi_class="multinomial",
        random_state=42
    ))
])

logreg.fit(X_train_tab, y_train)

print("Baseline LogisticRegression обучена.")
```

#### Основная модель: MiniROCKET

Подготовка данных для sktime

MiniROCKET ожидает 3D-тензор:

```python
(samples, channels, time_steps)
```

```python
X_train_ts = np.transpose(X_train, (0, 2, 1))
X_val_ts = np.transpose(X_val, (0, 2, 1))
X_test_ts = np.transpose(X_test, (0, 2, 1))

print(X_train_ts.shape)
```

```python
ts_mean = X_train_ts.mean(axis=(0, 2), keepdims=True)
ts_std = X_train_ts.std(axis=(0, 2), keepdims=True)

X_train_ts = (X_train_ts - ts_mean) / ts_std
X_val_ts = (X_val_ts - ts_mean) / ts_std
X_test_ts = (X_test_ts - ts_mean) / ts_std
```

##### Обучение MiniROCKET

```python
from sktime.classification.kernel_based import RocketClassifier

rocket_clf = RocketClassifier(
    num_kernels=5000,
    random_state=42
)

rocket_clf.fit(X_train_ts, y_train)

print("MiniROCKET classifier обучен.")
```

##### Первичная проверка качества

```python
from sklearn.metrics import accuracy_score

# baseline
y_pred_logreg = logreg.predict(X_test_tab)

# MiniROCKET
y_pred_rocket = rocket_clf.predict(X_test_ts)

acc_logreg = accuracy_score(y_test, y_pred_logreg)
acc_rocket = accuracy_score(y_test, y_pred_rocket)

print(f"LogisticRegression accuracy : {acc_logreg:.4f}")
print(f"MiniROCKET accuracy         : {acc_rocket:.4f}")
```

##### Визуальное сравнение моделей

```python
comparison_df = pd.DataFrame({
    "model": ["LogisticRegression", "MiniROCKET"],
    "accuracy": [acc_logreg, acc_rocket]
})

plt.figure(figsize=(6, 4))

sns.barplot(
    data=comparison_df,
    x="model",
    y="accuracy"
)

plt.title("Сравнение accuracy моделей классификации")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

plt.show()
```

### Выводы §2.3

В качестве baseline-подхода использована LogisticRegression на агрегированных статистиках временного окна. Такой подход прост и интерпретируем, однако частично теряет информацию о временной динамике

В качестве основной модели выбрана специализированная TSC-модель MiniROCKET. Она работает непосредственно с многомерными временными рядами и извлекает динамические паттерны климатических режимов

Для классификации использовались 30-дневные окна наблюдений, содержащие температуру, влажность, давление, ветер и осадки

Временное разделение train/validation/test выполнено корректно без утечки будущих данных

Предварительная проверка показывает, что MiniROCKET превосходит baseline-модель по accuracy, что подтверждает важность учёта временной структуры метеоданных при классификации климатов

В данной секции проводится подробная оценка качества моделей классификации климатических типов

Анализируются:

- Accuracy
- Precision / Recall / F1-score
- ROC-AUC
- матрицы ошибок
- сезонные ошибки классификации
- наиболее трудноразличимые климатические режимы
- важность признаков

Основное внимание уделяется модели MiniROCKET как специализированному методу классификации временных рядов

##### Accuracy / Precision / Recall / F1-score

```python
print("=== MiniROCKET classification report ===\n")

print(
    classification_report(
        y_test,
        y_pred_rocket,
        target_names=le.classes_
    )
)
```

Для многоклассовой задачи используется схема One-vs-Rest

```python
from sklearn.metrics import roc_auc_score

y_proba_rocket = rocket_clf.predict_proba(X_test_ts)

roc_auc = roc_auc_score(
    y_test,
    y_proba_rocket,
    multi_class="ovr",
    average="macro"
)

print(f"ROC-AUC OVR (macro): {roc_auc:.4f}")
```

## §2.5. Построение моделей прогнозирования

В данной секции реализуется второй этап системы — прогнозирование температуры на 30 дней вперёд для каждого города отдельно

Для каждого города строится собственная модель прогнозирования, поскольку различные климатические зоны обладают разной динамикой:

- континентальный климат
- морской климат
- субтропический климат
- муссонный климат

### Теоретическое обоснование выбора моделей

#### Почему для разных климатов могут быть оптимальны разные модели

Метеорологические ряды существенно отличаются между климатическими зонами:

| Климат | Особенности |
|--------|-------------|
| Континентальный | высокая сезонная амплитуда |
| Морской | сглаженные температурные колебания |
| Субтропический | слабая зимняя вариативность |
| Муссонный | выраженная нелинейность и резкие сезонные переходы |

Из-за этого единая модель для всех городов может работать хуже специализированных моделей.

### Какие модели плохо подходят для задачи

#### 1. Линейная регрессия

Не подходит для месячного прогноза:

- не моделирует сезонность
- плохо работает с нелинейностями
- не учитывает долгосрочную память ряда

#### 2. Простые autoregression-модели малого порядка

На горизонте 30 дней:

- быстро накапливают ошибку
- плохо моделируют годовую сезонность
- нестабильны при экстремумах

#### 3. KNN-регрессия

Для временных рядов:

- плохо экстраполирует
- чувствительна к шумам
- деградирует на длинных горизонтах

### Выбранные модели прогнозирования

Используются четыре подхода:

| Модель | Назначение |
|--------|------------|
| Seasonal Naive | климатический baseline |
| SARIMA | сезонная статистическая модель |
| Prophet | аддитивная модель сезонности |
| LightGBM | нелинейная ML-модель |

```python
TARGET_COL = "temperature_2m"

FORECAST_HORIZON = 30
LOOKBACK_DAYS = 30
```

```python
TRAIN_END = "2023-12-31"
VAL_END = "2024-12-31"

train_data = {}
val_data = {}
test_data = {}

for city in CITIES:

    df = daily_clean[city]

    train_data[city] = df.loc[:TRAIN_END]
    val_data[city] = df.loc["2024-01-01":VAL_END]
    test_data[city] = df.loc["2025-01-01":]

print("Данные разделены.")
```

##### 1. Seasonal Naive baseline

Baseline-модель использует климатическую норму.

```python
def seasonal_naive_forecast(
    df,
    origin,
    horizon=30
):

    preds = []

    for h in range(1, horizon + 1):

        target_date = origin + pd.Timedelta(days=h)

        doy = target_date.dayofyear

        hist = df[df.index.dayofyear == doy]

        preds.append(hist[TARGET_COL].mean())

    return np.array(preds)
```

##### 2. SARIMA

SARIMA моделирует:

- автокорреляцию
- сезонность
- долгосрочную структуру ряда

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_models = {}

for city in CITIES:

    series = train_data[city][TARGET_COL]

    model = SARIMAX(
        series,
        order=(2, 0, 2),
        seasonal_order=(1, 1, 1, 365),
        enforce_stationarity=False,
        enforce_invertibility=False,
    )

    res = model.fit(disp=False)

    sarima_models[city] = res

    print(f"SARIMA обучена: {city}")
```

##### 3. Prophet

Prophet хорошо моделирует:

- годовую сезонность
- плавные тренды
- календарные эффекты

```python
PROPHET_AVAILABLE = True

try:

    from prophet import Prophet

except ImportError:

    PROPHET_AVAILABLE = False

    print("Prophet не установлен.")
```

```python
prophet_models = {}

if PROPHET_AVAILABLE:

    for city in CITIES:

        dfp = (
            train_data[city][[TARGET_COL]]
            .reset_index()
            .rename(columns={
                "date": "ds",
                TARGET_COL: "y"
            })
        )

        model = Prophet(
            yearly_seasonality=True,
            weekly_seasonality=True,
            daily_seasonality=False,
        )

        model.fit(dfp)

        prophet_models[city] = model

        print(f"Prophet обучена: {city}")
```

```python
def prophet_forecast(
    model,
    origin,
    horizon=30
):

    future = pd.DataFrame({
        "ds": pd.date_range(
            origin + pd.Timedelta(days=1),
            periods=horizon,
            freq="D"
        )
    })

    fc = model.predict(future)

    return fc["yhat"].values
```

##### 4. LightGBM

LightGBM использует:

- лаговые признаки
- rolling statistics
- календарные признаки

```python
def build_lgb_features(df):

    data = df.copy()

    for lag in [1, 2, 3, 7, 14, 30]:

        data[f"lag_{lag}"] = (
            data[TARGET_COL].shift(lag)
        )

    for win in [3, 7, 14, 30]:

        data[f"roll_mean_{win}"] = (
            data[TARGET_COL]
            .rolling(win)
            .mean()
        )

        data[f"roll_std_{win}"] = (
            data[TARGET_COL]
            .rolling(win)
            .std()
        )

    data["month"] = data.index.month
    data["dayofyear"] = data.index.dayofyear

    return data
```

```python
import lightgbm as lgb

lgb_models = {}
lgb_feature_cols = {}

for city in CITIES:

    feat_df = build_lgb_features(
        train_data[city]
    )

    feat_df["target"] = (
        feat_df[TARGET_COL]
        .shift(-FORECAST_HORIZON)
    )

    feat_df = feat_df.dropna()

    feature_cols = [
        c for c in feat_df.columns
        if c not in [TARGET_COL, "target"]
    ]

    Xtr = feat_df[feature_cols]
    ytr = feat_df["target"]

    model = lgb.LGBMRegressor(
        n_estimators=300,
        learning_rate=0.03,
        max_depth=6,
        random_state=42
    )

    model.fit(Xtr, ytr)

    lgb_models[city] = model
    lgb_feature_cols[city] = feature_cols

    print(f"LightGBM обучена: {city}")
```

```python
def lgb_forecast(
    city,
    origin
):

    df = build_lgb_features(
        daily_clean[city]
    )

    row = df.loc[[origin]]

    X = row[lgb_feature_cols[city]]

    pred = lgb_models[city].predict(X)[0]

    return np.repeat(pred, FORECAST_HORIZON)
```

Для честной оценки используются множественные стартовые точки прогноза

```python
test_origins = pd.date_range(
    "2025-01-01",
    "2025-12-01",
    freq="MS"
)

print(test_origins)
```

```python
def rolling_forecast(
    city,
    model_name,
    origins,
    horizon=30
):

    rows = []

    for origin in origins:

        if model_name == "naive":

            preds = seasonal_naive_forecast(
                daily_clean[city],
                origin,
                horizon
            )

        elif model_name == "sarima":

            preds = sarima_models[city].forecast(horizon)

        elif model_name == "prophet":

            preds = prophet_forecast(
                prophet_models[city],
                origin,
                horizon
            )

        elif model_name == "lgbm":

            preds = lgb_forecast(
                city,
                origin
            )

        else:
            raise ValueError(model_name)

        future_dates = pd.date_range(
            origin + pd.Timedelta(days=1),
            periods=horizon,
            freq="D"
        )

        actual = (
            daily_clean[city][TARGET_COL]
            .reindex(future_dates)
            .values
        )

        for h, (d, yhat, ytrue) in enumerate(
            zip(future_dates, preds, actual),
            start=1
        ):

            rows.append({
                "city": city,
                "model": model_name,
                "origin": origin,
                "horizon": h,
                "date": d,
                "y_true": ytrue,
                "y_pred": yhat,
            })

    return pd.DataFrame(rows)
```

##### Формирование общей таблицы прогнозов

```python
all_forecasts = []

MODELS_TO_RUN = [
    "naive",
    "sarima",
    "lgbm",
]

if PROPHET_AVAILABLE:
    MODELS_TO_RUN.append("prophet")

for city in CITIES:

    for model_name in MODELS_TO_RUN:

        fc = rolling_forecast(
            city,
            model_name,
            test_origins,
            FORECAST_HORIZON
        )

        all_forecasts.append(fc)

forecast_long = pd.concat(
    all_forecasts,
    ignore_index=True
)

print(forecast_long.shape)

forecast_long.head()
```

##### Визуализация примера прогноза

## §2.6. Оценка качества прогнозирования

В данной секции проводится оценка качества моделей прогнозирования температуры

Анализируются:

- MAE
- RMSE
- MAPE
- WAPE
- Directional Accuracy
- остатки моделей
- распределение ошибок
- зависимость ошибки от горизонта прогноза

##### Метрики качества прогнозирования

```python
def _mae(a, p):

    return float(
        np.mean(np.abs(a - p))
    )


def _rmse(a, p):

    return float(
        np.sqrt(np.mean((a - p) ** 2))
    )


def _mape(a, p):

    eps = 1e-3

    return float(
        np.mean(
            np.abs((a - p) / (np.abs(a) + eps))
        ) * 100
    )


def _wape(a, p):

    return float(
        np.sum(np.abs(a - p))
        /
        (np.sum(np.abs(a)) + 1e-9)
        * 100
    )


def _dir_acc(a, p):

    da = np.diff(a)
    dp = np.diff(p)

    return float(
        np.mean(
            np.sign(da) == np.sign(dp)
        ) * 100
    )
```

##### Агрегирование метрик

```python
def group_metrics(df):

    a = df["y_true"].values
    p = df["y_pred"].values

    mask = (
        ~np.isnan(a)
        &
        ~np.isnan(p)
    )

    if mask.sum() < 2:

        return pd.Series({
            "MAE": np.nan,
            "RMSE": np.nan,
            "MAPE": np.nan,
            "WAPE": np.nan,
            "DirAcc": np.nan,
            "n": int(mask.sum())
        })

    a = a[mask]
    p = p[mask]

    return pd.Series({
        "MAE": _mae(a, p),
        "RMSE": _rmse(a, p),
        "MAPE": _mape(a, p),
        "WAPE": _wape(a, p),
        "DirAcc": _dir_acc(a, p),
        "n": int(mask.sum())
    })
```

##### Сводная таблица качества

```python
metrics_city_model = (
    forecast_long
    .groupby(["city", "model"])
    .apply(group_metrics)
    .reset_index()
)

metrics_city_model = (
    metrics_city_model
    .sort_values(["city", "MAE"])
)

metrics_city_model.round(3)
```

##### Анализ остатков

Для лучших моделей выполняются:

- тест Шапиро-Уилка
- тест Льюнга-Бокса
- тест Уайта

```python
from scipy.stats import shapiro

from statsmodels.stats.diagnostic import (
    acorr_ljungbox,
    het_white,
)
```

```python
def residual_tests(city, model_name):

    sub = forecast_long[
        (forecast_long["city"] == city)
        &
        (forecast_long["model"] == model_name)
    ]

    resid = (
        sub["y_true"] - sub["y_pred"]
    ).dropna().values

    if len(resid) < 30:

        return {"n": len(resid)}

    # Shapiro-Wilk
    sw_stat, sw_p = shapiro(
        resid[:5000]
    )

    # Ljung-Box
    lb = acorr_ljungbox(
        resid,
        lags=[10],
        return_df=True
    )

    lb_p = float(
        lb["lb_pvalue"].iloc[0]
    )

    # White test
    sub2 = sub.dropna(
        subset=["y_true", "y_pred"]
    ).copy()

    sub2["resid"] = (
        sub2["y_true"]
        -
        sub2["y_pred"]
    )

    try:

        X_white = np.column_stack([
            np.ones(len(sub2)),
            sub2["horizon"].values
        ])

        white = het_white(
            sub2["resid"].values,
            X_white
        )

        white_p = float(white[1])

    except Exception:

        white_p = np.nan

    return {
        "n": len(resid),
        "Shapiro_W": float(sw_stat),
        "Shapiro_p": float(sw_p),
        "LjungBox_p": lb_p,
        "White_p": white_p,
    }
```

```python
residual_summary = []

for city in CITIES:

    best_model = (
        best_model_per_city
        .loc[city, "model"]
    )

    tests = residual_tests(
        city,
        best_model
    )

    tests["city"] = city
    tests["best_model"] = best_model

    residual_summary.append(tests)

residual_df = (
    pd.DataFrame(residual_summary)
    .set_index("city")
)

residual_df
```

## 2.7. Интеграция двух этапов в единый пайплайн

### Постановка задачи

На предыдущих этапах были отдельно реализованы:

- модель классификации климатического типа / города по временному окну метеоданных
- набор моделей прогнозирования температуры на 30 дней вперёд для каждого города

На данном этапе требуется объединить оба компонента в единую систему, которая:

1. принимает на вход окно последних наблюдений
2. автоматически определяет наиболее вероятный город / климат
3. выбирает соответствующую модель прогнозирования
4. возвращает прогноз температуры на следующий месяц

### Общая схема пайплайна

Работа системы организована следующим образом:


```python
# Последние 30 дней наблюдений
#            ↓
# Этап 1: классификация климата / города
#            ↓
# Выбор лучшей модели для данного города
#            ↓
# Этап 2: прогноз температуры на 30 дней
#            ↓
# Финальный прогноз
```

```python
from dataclasses import dataclass
import numpy as np
import pandas as pd
```

```python
WINDOW_DAYS = 30
FORECAST_HORIZON = 30
```

```python
@dataclass
class MeteoPipeline:

    # классификатор
    classifier: object

    # параметры нормализации
    ts_channels: list
    ts_mean: np.ndarray
    ts_std: np.ndarray

    # кодировщик классов
    label_encoder: object

    # соответствие город -> климат
    climate_map: dict

    # лучшие модели прогнозирования
    best_model_per_city: pd.DataFrame

    # обученные модели
    sarima_models: dict
    prophet_models: dict
    lgb_forecasters: dict

    horizon: int = FORECAST_HORIZON
```

Классификатор временных рядов получает данные в формате:

```python
(batch, channels, time)
```

```python
def _build_ts_tensor(self, window_df: pd.DataFrame) -> np.ndarray:

    arr = (
        window_df[self.ts_channels]
        .values
        .T
        .astype(np.float32)
    )

    arr = arr[np.newaxis, ...]

    # нормализация
    arr = (arr - self.ts_mean) / self.ts_std

    return arr
```

##### Основной метод predict

```python
def predict(self, window_df: pd.DataFrame) -> dict:
    """
    На вход:
        DataFrame последних наблюдений

    На выход:
        город,
        климат,
        вероятности классов,
        модель,
        прогноз температуры
    """

    assert len(window_df) >= WINDOW_DAYS

    # последние WINDOW_DAYS дней
    w = window_df.iloc[-WINDOW_DAYS:].copy()

    # ==================================================
    # ЭТАП 1 — КЛАССИФИКАЦИЯ
    # ==================================================

    X_cls = self._build_ts_tensor(w)

    proba = self.classifier.predict_proba(X_cls)[0]

    pred_idx = int(np.argmax(proba))

    predicted_city = (
        self.label_encoder
        .inverse_transform([pred_idx])[0]
    )

    predicted_climate = self.climate_map[predicted_city]

    class_proba = {
        c: float(p)
        for c, p in zip(self.label_encoder.classes_, proba)
    }

    # ==================================================
    # ЭТАП 2 — ВЫБОР МОДЕЛИ ПРОГНОЗА
    # ==================================================

    chosen_model = (
        self.best_model_per_city
        .loc[predicted_city, "model"]
    )

    origin = window_df.index[-1]

    # ==================================================
    # ПРОГНОЗИРОВАНИЕ
    # ==================================================

    if chosen_model == "sarima":

        res = self.sarima_models[predicted_city]

        forecast = sarima_forecast_at(
            res,
            origin,
            self.horizon
        )

    elif chosen_model == "prophet":

        forecast = prophet_forecast_at(
            self.prophet_models[predicted_city],
            origin,
            self.horizon
        )

    elif chosen_model == "lgbm":

        forecast = lgb_forecast_at(
            predicted_city,
            origin,
            self.horizon
        )

    else:

        forecast = seasonal_naive_forecast(
            predicted_city,
            origin,
            self.horizon
        )

    # даты прогнозного горизонта
    future_dates = pd.date_range(
        origin + pd.Timedelta(days=1),
        periods=self.horizon,
        freq="D"
    )

    forecast_series = pd.Series(
        forecast,
        index=future_dates,
        name="temperature_forecast"
    )

    return {
        "predicted_city": predicted_city,
        "predicted_climate": predicted_climate,
        "class_proba": class_proba,
        "model_used": chosen_model,
        "forecast": forecast_series,
        "origin": origin,
    }
```

##### Создание объекта пайплайна

```python
pipeline = MeteoPipeline(
    classifier=rocket_clf,

    ts_channels=ts_channels,
    ts_mean=ts_mean,
    ts_std=ts_std,

    label_encoder=le,

    climate_map=CLIMATE_GROUPS,

    best_model_per_city=best_model_per_city,

    sarima_models=sarima_models,
    prophet_models=prophet_models,
    lgb_forecasters=lgb_forecasters,
)
```

#### Поведение моделей

**Prophet**

Лучше всего работает в городах с выраженной годовой сезонностью:
  - Москва
  - Санкт-Петербург
  - Сочи
  - Находка
Модель хорошо описывает:
  - долгосрочный сезонный цикл;
  - плавные климатические изменения.

**LightGBM**

Показывает лучшие результаты в более сложных климатических режимах:
  - Благовещенск
Причина:
  - нелинейные зависимости
  - резкие переходы между сезонами
  - влияние дополнительных ковариат

**SARIMA**

Показывает стабильные, но не лучшие результаты:
  - хорошо моделирует сезонные компоненты

**Seasonal Naive**

Остаётся сильным baseline для стабильных климатов:
  - Геленджик

#### Анализ остаточных ошибок

**Шапиро–Уилк**
 - `p-value < 0.05`
Нормальность остатков отвергается практически для всех моделей.
Причина:
  - экстремальные погодные события;
  - тяжёлые хвосты распределения ошибок.

**Льюнг–Бокс**

Обнаружена остаточная автокорреляция ошибок
Это ожидаемо, поскольку:
  - прогнозные окна перекрываются
  - соседние дни зависят друг от друга

**Тест Уайта**

Во многих случаях наблюдается гетероскедастичность:
 - `variance(error) ↑ with forecast horizon`
- То есть ошибка растёт с увеличением горизонта прогноза
